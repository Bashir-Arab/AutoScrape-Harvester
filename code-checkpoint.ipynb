```python
# open a page
def getStarted():
    # open the page
    try:
        options = web.ChromeOptions()
        #options.add_argument("--headless")
        options.add_argument("--silent")
        options.add_argument("--disable-gpu")
        options.add_argument("--log-level=3")
        options.add_argument('--disable-blink-features=AutomationControlled')
        options.add_argument('--disable-software-rasterizer')
        options.add_experimental_option("excludeSwitches", ["enable-automation"])
        options.add_experimental_option('useAutomationExtension', False)
        options.add_argument('--ignore-certificate-errors')
        options.add_argument('--ignore-ssl-errors')
        options.add_experimental_option('excludeSwitches', ['enable-logging'])
        driver = web.Chrome(options=options, executable_path= r'C:\Users\omran\OneDrive\Coding\bot for amazon\chromedriver.exe')
        driver.execute_cdp_cmd('Network.setUserAgentOverride', {"userAgent": 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/83.0.4103.53 Safari/537.36'})
        driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
        path = input("Enter URL: ")
        driver.get(path)
        driver.minimize_window()
        return driver
    except:
        print('Driver Error')
        return None
        
```

```python
# extract all the comments
def extract_comments(driver):

    data = []
    button = driver.find_elements_by_link_text('اطلع على جميع المراجعات')
    num = len(button)
    if(num >= 1): button[0].click()
    else: return data

    comments = len(driver.find_elements_by_xpath("/html/body/div[1]/div[3]/div/div[1]/div/div[1]/div[5]/div[3]/div/div"))
    print("Number of comments: ", comments)
    try:

        for i in range(1000):

            for i in range(2,comments):
                b = driver.find_elements_by_xpath("/html/body/div[1]/div[3]/div/div[1]/div/div[1]/div[5]/div[3]/div/div["+str(i)+"]/div/div/div[4]")
                data.append(b[0].text)
                print("B: ", b)
                print("comment: ", b[0].text)
                print('______________________________________________________________________')

            button = driver.find_element_by_xpath('//*[@id="cm_cr-pagination_bar"]/ul/li[2]/a')
            button.click()
            time.sleep(1)

    except BaseException as e:
        print(e, '____DONE____')
        return data
```

```python
# start the program 
driver = getStarted()
driver.maximize_window()
data = []
```

```python
driver.find_elements_by_xpath("/html/body/div[1]/div[2]/div[1]/div[1]/div/span[3]/div[2]/div[4]")[0].text
```

```python
# open the products page

for i in range(1000):

    products = len(driver.find_elements_by_xpath('/html/body/div[1]/div[2]/div[1]/div[1]/div/span[3]/div[2]/div'))

    if(products == 0):
        break
    else:
        if(products == 27): s = 2
        else: s = 1
        print(products)
        for j in range(s,(products-1)):
            product = driver.find_elements_by_xpath("/html/body/div[1]/div[2]/div[1]/div[1]/div/span[3]/div[2]/div["+str(j)+"]")
            if(product[0].text == ".قد تختلف الأسعار والتفاصيل الأخرى حسب حجم المنتج ولونه"):
                continue
            else:
                product = driver.find_elements_by_xpath("/html/body/div[1]/div[2]/div[1]/div[1]/div/span[3]/div[2]/div["+str(j)+"]/div/div/div/div/div/div[1]")
                time.sleep(5)
                newTap = ActionChains(driver) \
                    .key_down(Keys.CONTROL) \
                    .click(product[0]) \
                    .key_up(Keys.CONTROL) \
                    .perform()
                driver.switch_to.window(driver.window_handles[1])
                data.extend(extract_comments(driver))
                driver.close()
                driver.switch_to.window(driver.window_handles[0])

    driver.find_element_by_class_name('a-last').click() # next_page
    time.sleep(2)
```
